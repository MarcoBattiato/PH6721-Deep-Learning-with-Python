## Python

Python is a modern, robust, high level programming language. It is very easy to pick up even if you are completely new to programming. Python is an interpreted, high-level, general-purpose programming language with very extensive libraries. It also supports multiple programming paradigms.

**Interpreted**<br>
Languages like Fortran, C++, or Java require the code to be compiled. That means that after the entire code is written, the so-called compiler has to translate the code from the programming language into machine commands (note that a single instruction can correspond to very many machine instructions) before it can be executed. Performing this step takes  time. However while translating, the compiler can optimise the code in several ways (actually most of the compilation time is spent in this optimisation process). The optimisation can speed up the code by several orders of magnitude (especially if some techniques are adopted to maximally leverage the optimization). In particular the compiler could use its knowledge of the full program to optimise it.

Languages like python, or Matlab are instead interpreted. They do not have a compiler, but an interpreter. The interpreter simply follows the program flow instruction by instruction (a bit how you would do when you try to understand what a program does), reads the instruction and executes the binary code associated to it, without trying to use any knowledge of what came before or what might come after to optimize the execution. Notice that the binary code for each single instruction has been optimised when the interpreter was constructed. This is good news as you don't need to wait for this when you want your program to run. The drawback is that the optimisation is limited as it has to work in the most generic situation possible.

This has advanteges and disadvantages. 

The main advantage of interpreted languages is that the code can be executed in pieces, which allows for a much easier human overview of the program workflow. This is great for debugging and testing. The importance of this advantage is not to be underestimated, since these two steps are in most cases where the biggest part of human time is spent.
Being an interpreted language also removes the need to take the further step of compiling to construct an executable (this step takes time as the compiler needs time to analyse the code and figure out how to make it faster). This is usually advantageous in situations where the program is not computationally intensive.
Finally, as there is no executable, the written program is very easy to transfer from one platform to the other. It is very appreciated in companies that provide taylor made programs to other companies, as the client can immediately put the program to use without the requirement of several technical steps which the final user might struggle with.

The disadvantages are that the code runs slower than for compiled languages. The difference however can vary enormously depending on what one is actually doing. Simpler programs usually tend to perform well even in python, while complex programs or algorithms can heavily suffer when executed in interpreted languages. However one always has to balance the cost of computer time (usually cheap) vs human time (usually expensive).
Another disadvantage has again to do with the lack of a compiler. Another important role of a compiler is to check for errors. Of course the compiler can find only certain types of errors, but good programming practices can make the compiler's error-checking very powerful. Notice that this error-checking happens before the program runs. In interpreted languages this error checking can be done only on the instruction that is currently running. This means that to find an error towards the end of the program you have to run almost the whole program before finding it. Again this is not an issue in programs that are not computationally expensive.

**High-level**<br>
Being high level refers to a number of characteristics that might not be very easy to understand if you are not familiar with lower level languages. Sufficies to say that python tries to make as many technical choices for you as it can (for instance memory management, typing, dispatch, ...). Of course those choises might not be the best in your specific case: they are designed to work well in _most cases_, but with the constraint that they work in _every_ case.
The great advantage of being high level is that writing programs in Python is very quick!

**Rich and general-purpose**

Unlike Matlab, that is specifically built to handle matrix operations, python is built with the aim of being a convenient language to work in very different fields: from physics, to databases, AI, GUI, communication, etc. This means that choices have been done to make everyone as comfortable as possible.
This made python a very popular and widespread language. However we must be careful, as some mathematical operations might not immediately translate into python code (and actually the pyhton code you might be tempted to write does not do what you hope it will).

The great thing about python is that it comes with a very large collection of libraries for everything from scientific computing to web services. As one of the most important rules in programming is "do not reinvent the wheel" we will benefit greatly from using these libraries.

**Multi-paradigm**<br>
The aims of programming are multiple. Of course we need our programs to (mostly) work. In some industries perfect compliance with specifications is not an absolute must (it will be in this course, sorry ;) ). Probably the most famous example is the number of crashes that are acceptable in widely used operating systems.
Other aspects are often as (and sometimes more) important: development time, code readability, and code reusability.
Programming paradigms are developed to address these problems. One key element in the story is abstraction. Python caters for object oriented and functional programming with module system that allows large and complex applications to be developed in Python. <br>
In this course we will learn some very basic concepts of object-oriented programming. We will also learn some basics of good programming practice.

## Control Flow Statements

Before introducing the control flow statements it is important to understand a key feature about Python. 

## Indentation and blocks of code

First let's clarify what is meant by **indentation**. Indentation is simply the amount of empty spaces present in each line before the actual instruction starts.

Now that we know what indentation is, we can proceed. We will often find ourselves in the position of having to indicate a block of instructions. This happens for instance when controlling the flow of the program. Let us suppose that under one condition we want python to execute a certain set of instructions otherwise another set. A set of instructions is called a **block** of code. To achieve that we need to be able to distinguish the different blocks.

Certain languages allow us to specify blocks of code by enclosing them within, for instance, curly brackets. 
```
if condition then 
{ instruction1
instruction2
instruction3}
else 
{ instruction4
instruction5
instruction6}
```
However it became extremely common to also increase the indentation of the code when starting a new block, and reduce it when the block is finished. 
```
if condition then { 
    instruction1
    instruction2
    instruction3
} else { 
    instruction4
    instruction5
    instruction6
}
```
Indentation to highlight blocks is used as it makes the program structure very visually evident, much more than the use of brackets. That is why in many languages, even though indentation is not required as part of the syntax, it is almost ubiquitously used, as it makes the code much much easier to read.

Python took this a step forward, making indentation to define blocks compulsory and the brackets not necessary. This is a good idea as it forces programmers to adhere to good programming standards. But that also requires us to be careful with the indentation, as adding a wrong number of spaces might change what the interpreter understands and therefore what the program does.

We will see how blocks of codes are identified, while studying control flow statements.

## Conditionals

### If

The ``if`` control flow statement triggers the execution of a block of code if a certain condition is met. More specifically the condition has to be an expression (notice that an expression can be even a single variable) that has a boolean value.
```python
if some_condition:
    instruction_block_to_be_executed_if_some_condition_is_true
```

````python
userReply = input("Do you like python? ")
if userReply == "yes":
    print("That makes me very happy!")
````

Let's now see what to do if we want to execute more instructions.

````python
userReply = input("Do you like python? ")
if userReply == "yes":
    print("That", end = '')        # Notice how we can prevent python from adding a new line after a print
    print(" makes me", end = '')
    print(" very happy!")
print("This program ends here")
````

You might have noticed that if the condition is met, then the block of code is executed. If the condition is not met the entire block is skipped. Notice how the istructions that have a reduced indentation are executed every time because they do not belong to the block of code.

To make our program a bit more interesting we can use a different control flow statement:

### If-else

```python
if some_condition:
   instruction_block_to_be_executed_if_some_condition_is_true
else:
   instruction_block_to_be_executed_if_some_condition_is_false
```
    

````python
userReply = input("Do you like python? ")
if userReply == "yes":
    print("That", end = '')        # Notice how we can prevent python from adding a new line after a print
    print(" makes me", end = '')
    print(" very happy!")
else:
    print("That", end = '')        
    print(" makes me", end = '')
    print(" very sad...")
print("This program ends here")
````

The program above is however not fully satisfying. We can use another type of control flow to make it better.

### Else if

```python
if condition1:  
    instruction_block_to_be_executed_if_condition1_is_true
elif condition2:
    instruction_block_to_be_executed_if_condition1_is_false_but_condition2_is_true
else:
    instruction_block_to_be_executed_if_both_condition1_and_condition2_are_false
```

````python
userReply = input("Do you like python? ")
if userReply == "yes":
    print("That", end = '')        # Notice how we can prevent python from adding a new line after a print
    print(" makes me", end = '')
    print(" very happy!")
elif userReply == "no":
    print("That", end = '')        
    print(" makes me", end = '')
    print(" very sad...")
else:
    print("I don't understand what you just said")       
print("This program ends here")
````

All the control flow statements can be put inside other control flow statements. In that case, they are called nested if statements

````python
userReply = input("Do you like python? ")
if userReply == "yes":
    print("That", end = '')        # Notice how we can prevent python from adding a new line after a print
    print(" makes me", end = '')
    print(" very happy!")
elif userReply == "no":
    userReply = input("Seriously? ")
    if userReply == "yes":
        print("That makes me very sad...")
    else:
        print("I hoped you would say that!")
else:
    print("I don't understand what you just said")       
print("This program ends here")
````

## Loops

### For loops

A `for` loop repeats a block of code usually a specified number of times. Depending on the programming language you are accustomed to, `for` loops can take some different flavors and functionalities. Here we will learn how they work in python.
```python
for variable in something:
    algorithm
```

Python requires you to specify a variable and a list of values that variables will assume. Then python will execute the block below several times, by assigning to the given variable one of the values in the list (in order), until it has run over the entire list.

````python
for i in [0,1,2,3,4,5]:
    print(i)
print("-------")
for i in [4,2,6,1,5,6]:
    print(i)
print("-------")
for i in ["a","b","c","asdg"]:
    print(i)
print("-------")
````

When looping over integers the **range()** function is useful which generates a list of integers filled with the desired range:
* range(n) =  0, 1, ..., n-1
* range(m,n)= m, m+1, ..., n-1
* range(m,n,s)= m, m+s, m+2s, ..., m + ((n-m-1)//s) * s

````python
total = 0
for i in range(5):
    total += i
print("total =",total)
````

Notice that strings are treated as lists of characters.

````python
for ch in 'abc':
    print(ch)
````

# Functions

If you have had some experience with programming you might have noticed that it is common to have to give the same or similar sequences of instructions several times. One possibility would be to simply patiently rewrite everything every time. However apart from being very boring (and a waste of your time), we will see that it is a bad practice (more on this later).

To address these cases (almost) all languages offer the possibility of constructing functions. A programming function is a block of instructions to which a name has been assigned: this allow to simply call the name of the function and the interpreter will understand that it has to execute all the instructions contained within the function.

```python
def funcname():
    statement1
    statement2
    statement3
```

This is indeed a neat trick, but functions are much more flexible. Each function can be defined as accepting parameters from the final user. The instructions within the function will then perform the defined operations according to the passed parameters

```python
def funcname(arg1, arg2,... argN):
    statements_depending_on_arg1_arg2_etc
```

These types of functions are useful to execute some operations. However python functions offer another very interesting possibility: they can return a value. To this respect they can behave as mathematical functions, where the passed parameters are used to calculate the return value.

```python
def funcname(arg1, arg2,... argN):
    statements
    return <value>
```

Functions are very important, not only because they can save your time, but also because they provide a very powerful way to share code. Most programs tend to perform a large number of relatively complex operations. Separating the tasks and wrapping them in functions that can be used upon request is a very important way to manage complexity. This is because we can work a lot on solving a part of the problem we have, wrap it into a function and then simply request the solution without having to worry about or having to remember how it was calculated. This also allows your colleagues to use the function you programmed without having to spend the time to understand how it works.  

To achieve this it is important that every function has a clear behaviour and that that behaviour is well documented. For this python provides syntax to easily store documentation within the function definition:

```python
def funcname(arg1, arg2,... argN):
    ''' Document String'''
    statements
    return <value>
```

If a function provides as document string, it can be seen by calling
```python
help(funcname)
```
Notice that you can see the document string of most functions we will be using in python (for instance you can do `help(print)`).

We will learn what are good practices in producing documentation.

## Return Statement

When the function results in some value and that value has to be stored in a variable or needs to be sent back or returned for further operation to the main algorithm, a return statement is used.

````python
def times(x,y):
    z = x*y
    return z

c = times(4,5)
print(c)
````

## Objects, variables, and function arguments

If you are familiar with other programming languages, you might wonder whether Python passes function arguments *by value* or *by reference*.

For Python, this terminology is not very helpful. A better way to think about it is this:

> In Python, variables are names attached to objects.

When we pass a variable to a function, Python passes the object that the variable refers to. Inside the function, the parameter becomes another name for the same object.

This matters because some objects can be modified after they are created. These are called **mutable** objects. Lists and NumPy arrays are mutable.


### Two variables can refer to the same object

Consider the following example:

````python
x = [1, 2, 3]
y = x

y[0] = 100

print(x)
print(y)
````

Changing the list through `y` also changes what we see through `x`, because there is only one list object. Both variables refer to the same object.

If we want an independent list, we must explicitly create a copy.

````python
x = [1, 2, 3]
y = x.copy()

y[0] = 100

print(x)
print(y)
````

Now `x` and `y` refer to two different list objects.

### Modifying an object inside a function

The same idea applies to function arguments. If the object is mutable, the function can modify it.

````python
def modify_list(L):
    L[0] = 999

numbers = [1, 2, 3]

modify_list(numbers)

print(numbers)
````

The function did not receive a separate copy of the list. The parameter `L` referred to the same list object as `numbers`.

### Reassigning a variable is different

Now compare the previous example with this one:

````python
def replace_list(L):
    L = [999, 999, 999]

numbers = [1, 2, 3]

replace_list(numbers)

print(numbers)
````

Here the original list is unchanged. Inside the function, `L` is reassigned to a new list. This does not change what `numbers` refers to outside the function.

So the important distinction is:

- modifying an existing object can affect the caller;
- reassigning a local variable inside the function does not affect the caller.


### A dangerous example

Let us now look at a function that computes the sum of the squares of a list of numbers. The following version works numerically, but it has a serious problem.

````python
def sumSquares(vectorOfNumbers):
    for i in range(len(vectorOfNumbers)):
        vectorOfNumbers[i] *= vectorOfNumbers[i]    # This modifies the input object!
        
    summation = 0
    for elem in vectorOfNumbers:
        summation += elem
        
    return summation
    
print(sumSquares([2, 3, 5, 12, 4]))
````

The problem becomes visible when the input list already exists before calling the function.

````python
myListOfNumbers = [2, 3, 5, 12, 4]

print(sumSquares(myListOfNumbers))
print(myListOfNumbers)
````

The function has changed `myListOfNumbers`. This is usually not what we want from a function whose purpose is only to compute a value.

### Safer version: copy before modifying

If a function needs to modify the data internally, but the original object should remain unchanged, make an explicit copy.

````python
def sumSquares(vectorOfNumbers):
    localCopy = vectorOfNumbers.copy()     # Make an independent copy
    
    for i in range(len(localCopy)):
        localCopy[i] *= localCopy[i]
        
    summation = 0
    for elem in localCopy:
        summation += elem
        
    return summation

myListOfNumbers = [2, 3, 5, 12, 4]

print(sumSquares(myListOfNumbers))
print(myListOfNumbers)
````

Now the function returns the sum of squares, but the original list is not modified.

### NumPy arrays behave similarly

NumPy arrays are also mutable. This is very important in scientific computing and deep learning, because arrays and tensors are often large objects.

````python
import numpy as np

x = np.array([1, 2, 3])
y = x

y[0] = 100

print(x)
print(y)
````

Again, `x` and `y` refer to the same object. To create an independent array, use `.copy()`.

````python
x = np.array([1, 2, 3])
y = x.copy()

y[0] = 100

print(x)
print(y)
````

### Rule of thumb

- If your function is supposed to compute and return a value, avoid modifying its input arguments.
- If you need to modify an input internally, first make an explicit copy.
- If your function is supposed to modify an object, make that clear in the function name and documentation.
- Remember: assignment such as `y = x` does not copy the object. It only creates another name for the same object.

This distinction is especially important later, when working with NumPy arrays, Pandas DataFrames, and deep-learning tensors.


## Global and Local Variables

A variable created inside a function is usually **local**: it exists only while that function is running.

A variable created outside all functions is **global**: it can be read from many parts of the notebook.

For small examples, global variables can be convenient. In real programs and machine-learning notebooks, however, global state can easily create confusing bugs: a function may depend on a value that was changed in a different cell, or a notebook may stop working when cells are run in a different order.

A good rule is:

- use global variables mainly for constants, such as configuration values;
- pass information into functions through arguments;
- return results explicitly;
- avoid functions that silently modify global state.


````python
eg1 = [1,2,3,4,5]
````

In the below function we are appending an element to the declared list inside the function. eg2 variable declared inside the function is a local variable.

````python
def egfunc1():
    x=1
    def thirdfunc():
        x=2
        print("Inside thirdfunc x =", x) 
    thirdfunc()
    print("Outside x =", x)

egfunc1()
````

The example below shows why global variables should be used carefully. The keyword `global` tells Python that the function wants to modify the global variable `x`, rather than create a local variable called `x`.

This is sometimes useful, but it makes the function less self-contained and harder to reuse.

## Lambda Functions

A **lambda function** is a short anonymous function: a function defined directly where it is needed, rather than with a separate `def` block.

The syntax is:

```python
lambda arguments: expression
```

Lambda functions are useful when a function is needed only once, especially when another function expects a function as one of its arguments. This happens often in data analysis and deep-learning code, for instance when sorting, transforming datasets, or defining small preprocessing operations.


````python
square = lambda x: x**2

square(8)
````

The lambda function above is equivalent to the following ordinary function definition:

````python
def square(x):
    return x**2

square(8)
````

The main advantage of lambda functions is that they can be written directly inside another function call. For example, the function `sorted(...)` can receive a `key` function telling Python what criterion should be used for sorting.

````python
pairs = [(1, 3), (2, 1), (4, 2)]

# Sort the pairs according to their second element.
sorted(pairs, key=lambda x: x[1])
````

A good rule of thumb is:

- use `def` for functions that are important, reused, or need a clear name;
- use `lambda` for very short functions that are passed as arguments and used only locally.

Avoid writing complicated lambda functions. If the expression is difficult to read, define a normal function instead.
