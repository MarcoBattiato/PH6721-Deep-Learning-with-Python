### Question 1 — Indexing a list

`values = [10, 20, 30, 40, 50]`

What do `values[0]` and `values[-1]` print?

```python
values = [10, 20, 30, 40, 50]

print(values[0])
print(values[-1])
```

### Question 2 — Slicing a list

Using the same list, what does `values[1:4]` print? How many elements does it contain?

```python
print(values[1:4])
print(len(values[1:4]))
```

### Question 3 — A loop and a total

What does this print? Work out the arithmetic before running it.

```python
total = 0

for i in range(5):
    total += i

print(total)
```

### Question 4 — Writing a function

What does this print?

```python
def area(width, height):
    return width * height

print(area(3, 4))
print(area(height=10, width=2))
```

### Question 5 — A dictionary

What do the two lines print? What would happen if you asked for a key that is not there?

```python
masses = {"hydrogen": 1.008, "helium": 4.003, "lithium": 6.941}

print(masses["helium"])
print(len(masses))
```

### Question 6 — Formatting a string

What exactly does this print, including the number of decimal places?

```python
name = "helium"
mass = 4.002602

print(f"The mass of {name} is {mass:.2f}")
```

### Question 7 — A condition

What does this print? Notice that the list is looped over, not just tested once.

```python
for value in [3, 8, 12]:
    if value > 10:
        print(value, "is large")
    elif value > 5:
        print(value, "is medium")
    else:
        print(value, "is small")
```

### Question 8 — A list comprehension

What list does this produce? If the syntax is unfamiliar, that is worth knowing — it appears
throughout the course.

```python
squares = [v**2 for v in range(6) if v % 2 == 0]

print(squares)
```
