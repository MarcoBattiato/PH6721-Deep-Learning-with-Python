## 3.1 Separators and line endings

`print` accepts several values and puts a space between them, then moves to a new line.

Both behaviours can be changed:

- `sep` sets what goes *between* the values;
- `end` sets what goes *after* the last one. Setting `end=''` prevents the newline, so the next
  `print` continues on the same line.

````python
print("a", "b", "c")                 # default: separated by spaces
print("a", "b", "c", sep="")         # no separator
print("a", "b", "c", sep=" | ")      # a custom separator

print("no newline here -> ", end="")
print("...so this continues the same line")
````

## 3.2 f-strings

Very often we want to place the value of a variable inside a string.

The modern way to do this is an **f-string**: put an `f` immediately before the opening quote, and
then write any expression inside curly braces.

````python
name = "Ada"
year = 1843

print(f"{name} wrote the first published algorithm in {year}.")
````

Anything inside the braces is an expression, not just a variable name, so calculations work too.

````python
radius = 2.5
print(f"A circle of radius {radius} has area {3.14159 * radius**2}.")
````

## 3.3 Controlling how numbers are printed

The area above was printed with far more digits than we need. A **format specifier** after a colon
controls how the value appears.

| Specifier | Meaning |
|---|---|
| `{x:.3f}` | fixed point, 3 decimal places |
| `{x:.3e}` | scientific notation, 3 decimal places |
| `{x:8.3f}` | as above, padded to at least 8 characters wide |
| `{x:<8}` | left aligned in 8 characters |
| `{x:>8}` | right aligned in 8 characters |
| `{x:^8}` | centred in 8 characters |

The width is a *minimum*: a value that needs more room simply takes it.

````python
area = 19.6349375

print(f"default:      {area}")
print(f"3 decimals:   {area:.3f}")
print(f"scientific:   {area:.3e}")
print(f"width 12:     |{area:12.3f}|")
print(f"left in 12:   |{area:<12.3f}|")
print(f"centred in 12:|{area:^12.3f}|")
````

Aligning numbers to a fixed width is what makes a printed table readable. We will use this whenever
we report results.

````python
print(f"{'method':<12}{'accuracy':>10}")
print("-" * 22)
for method, accuracy in [("linear", 0.7241), ("boosting", 0.9033), ("neural net", 0.8817)]:
    print(f"{method:<12}{accuracy:>10.3f}")
````

# 5. Mathematical functions and importing libraries

Python's core language deliberately stays small. Mathematical functions such as `sin` and `exp`
live in **libraries** that we import when we need them.

The standard library includes a module called `math`, which is fine for single numbers. In this
course we will almost always use **NumPy** instead, because unlike `math` it also works on whole
arrays of numbers at once — which is what scientific computing needs.

There are four ways to import a library.

**1. Import everything into the current namespace**

```python
from numpy import *
sin(0.3)
```

This is **bad practice**. Libraries contain a large number of functions, we usually want only a
few of them, and we often do not even know what else is in there. All those names are now loaded
and can silently collide with names we define ourselves.

**2. Import the library itself**

```python
import numpy
numpy.sin(0.3)
```

Better, because it is explicit about where the function comes from, but rather verbose.

**3. Import the library under a short alias — the best option**

```python
import numpy as np
np.sin(0.3)
```

Short to write, and still explicit about the origin of the function. Before inventing an alias for
a well-known library, check what the established one is — `np` for NumPy, `pd` for pandas,
`plt` for matplotlib's pyplot. Using the conventional alias makes your code recognisable to
everyone else.

**4. Import one named thing from a library — also good practice**

```python
from sklearn.linear_model import LogisticRegression
```

This looks like option 1 and is not the same thing at all. Option 1 imports **everything**, under
unqualified names, without your knowing what arrived; that is what makes it dangerous. This form
imports **exactly the one name you asked for**, which is perfectly clear.

It is the standard way to use scikit-learn, whose tools live in submodules and where you almost
always want one specific object. You will see it constantly from Class 1 onward, so it is worth
recognising rather than mistaking for the starred import.

````python
import numpy as np

3 / 5 + np.sin(1.2)
````

Two things about that result. NumPy prints its own scalar type, so it appears as
`np.float64(1.532...)` rather than as a bare number — it is a floating-point number and behaves
exactly like one; the prefix is only NumPy naming the type it used.

And the fourth import form is worth seeing run, since `math` is a good small example of it.

````python
from math import sqrt, pi     # just these two names, not the whole module

print(sqrt(16))
print(pi)
````

### Slicing

A **slice** extracts a whole range of elements at once, written `list[start:end]`. The element at
`start` is included; the element at `end` is **not**. That looks arbitrary at first, but it makes
the number of elements returned simply `end - start`.

````python
values = [10, 20, 30, 40, 50, 60]

print(values[1:4])    # elements 1, 2 and 3 — element 4 is excluded
print(values[:3])     # from the beginning
print(values[3:])     # through to the end
print(values[:])      # the whole list
print(values[-2:])    # the last two
print(values[::2])    # every second element
````

Slicing a list always produces a **new, independent list**: changing the slice leaves the original
untouched. Fix that in your mind now, because `01B` shows that NumPy arrays deliberately behave the
opposite way, and the contrast between the two is a reliable source of bugs.

````python
list_a = [10, 20, 30, 40, 50]
list_b = list_a[1:4]

list_b[0] = 99

print("list_a:", list_a)   # unchanged — slicing a list copies
print("list_b:", list_b)
````

To find the length of the list or the number of elements in a list, **len( )** is used.

````python
len(numbers)
````

Three other built-in functions work on any list of numbers, and are worth knowing now because we
will use them shortly:

- `max(...)` — the largest element;

- `sum(...)` — the total.

````python
measurements = [2.4, 9.1, 5.5, 0.8]

print("largest: ", max(measurements))
print("smallest:", min(measurements))
print("total:   ", sum(measurements))
print("mean:    ", sum(measurements) / len(measurements))
````

Lists are not the only container Python offers. Three others appear constantly in scientific and
machine-learning code.

## 7.1 Tuples

A **tuple** is like a list, but it cannot be changed after it is created. Tuples are written with
round brackets instead of square ones.

Because they are fixed, tuples are used for things that naturally come as a small fixed group: a
coordinate, an RGB colour, the shape of an array.

````python
point = (2.0, 3.5)

print(point)
print(f"x = {point[0]}, y = {point[1]}")
````

````python
# This cell is intentionally wrong. Run it and inspect the error.
# Tuples are immutable: their elements cannot be reassigned.

point[0] = 99.0
````

### Unpacking

A tuple can be split into separate variables in one line. This is called **unpacking**, and it is
extremely common — in particular, functions that return several values return them as a tuple.

````python
point = (2.0, 3.5)

x, y = point          # unpacking

print(f"x = {x}")
print(f"y = {y}")
````

NumPy uses tuples for array shapes, which is why you will see `(150, 4)` rather than `[150, 4]`
when you ask an array how big it is. We will meet this in notebook `01B`.

## 7.2 Dictionaries

A **dictionary** stores pairs: a **key**, and the **value** associated with it. Where a list is
indexed by position, a dictionary is indexed by key.

Dictionaries are written with curly brackets, with `key: value` pairs separated by commas.

They are everywhere in machine learning: configuration settings, results tables, mappings from a
class number to a class name, and the objects that store the parameters of a trained model.

````python
melting_point = {
    "hydrogen": 14.01,
    "helium": 0.95,
    "lithium": 453.65,
}

print(melting_point["lithium"])
````

Values are read and written using the key in square brackets. Assigning to a key that does not
exist yet adds it.

````python
melting_point["carbon"] = 3823.0     # add a new entry
melting_point["helium"] = 0.9500     # overwrite an existing one

print(melting_point)
print(f"The dictionary now holds {len(melting_point)} elements.")
````

Asking for a key that is not there raises an error. When a key may legitimately be missing, use
`.get(key, default)` instead, which returns the default rather than failing.

To ask whether a key exists at all, use the `in` operator, which returns `True` or `False`. It
works on lists and sets too.

````python
# This cell is intentionally wrong. Run it and inspect the error.

melting_point["unobtainium"]
````

````python
print(melting_point.get("unobtainium", "not measured"))

print("carbon" in melting_point)        # membership test
print("unobtainium" in melting_point)
````

### Looping over a dictionary

Looping over a dictionary directly gives the keys. Use `.items()` to get keys and values together,
unpacked into two variables.

````python
for element in melting_point:
    print(element)

print("-" * 30)

for element, temperature in melting_point.items():
    print(f"{element:<10}{temperature:>10.2f} K")
````

## 7.3 Sets

A **set** is an unordered collection with no duplicates. Sets are useful for answering "which

# 9. Control flow

Before introducing the control flow statements it is important to understand a key feature about Python. 

## 9.1 Indentation and blocks of code

First let's clarify what is meant by **indentation**. Indentation is simply the amount of empty spaces present in each line before the actual instruction starts.

Now that we know what indentation is, we can proceed. We will often find ourselves in the position of having to indicate a block of instructions. This happens for instance when controlling the flow of the program. Let us suppose that under one condition we want python to execute a certain set of instructions otherwise another set. A set of instructions is called a **block** of code. To achieve that we need to be able to distinguish the different blocks.

Certain languages allow us to specify blocks of code by enclosing them within, for instance, curly brackets. 
```
if (condition) {
instruction1
instruction2
instruction3
} else {
instruction4
instruction5
instruction6
}
```
However it became extremely common to also increase the indentation of the code when starting a new block, and reduce it when the block is finished. 
```
if (condition) {
    instruction1
    instruction2
    instruction3
} else {
    instruction4
    instruction5
    instruction6
}
```
Indentation to highlight blocks is used as it makes the program structure very visually evident, much more than the use of brackets. That is why in many languages, even though indentation is not required as part of the syntax, it is almost ubiquitously used, as it makes the code much much easier to read.

Python took this a step forward, making indentation to define blocks compulsory and the brackets not necessary. This is a good idea as it forces programmers to adhere to good programming standards. But that also requires us to be careful with the indentation, as adding a wrong number of spaces might change what the interpreter understands and therefore what the program does.

We will see how blocks of codes are identified, while studying control flow statements.

Let us go back to our program asking the user whether he/she liked python. The program (apart from being of dubious utility) is still unsatisfactory. If the user does not give a meaningful answer, he/she does not have the chance to answer again. 
In orer to fix that we can use the `while` control statement.
```python
while some_condition:  
    instructions
```
Notice that the condition is tested at every iteration.

Here a `while` loop is genuinely the right tool, because we do not know in advance how many
attempts it will take. We simulate a sequence of replies rather than asking, so that the notebook
still runs from top to bottom.

````python
replies = ["maybe", "dunno", "yes"]   # pretend the user typed these, in order
attempt = 0

user_reply = replies[attempt]

while user_reply != "yes" and user_reply != "no":
    print(f"I did not understand '{user_reply}', please repeat.")
    attempt += 1
    user_reply = replies[attempt]

if user_reply == "yes":
    print("That makes me very happy!")
else:
    print("That makes me very sad...")
````

The loop above stops because one of the pretend replies is finally acceptable, and the condition it
was testing becomes false. It has a second backstop besides: it would run out of replies and raise
an `IndexError` rather than spin. A real `while` loop reading from a user has neither, and one whose condition never becomes false will run
until you stop it. If a cell seems stuck, use **Runtime → Interrupt execution**.

## 9.4 List comprehensions

A very common pattern is: start with a sequence, do something to each element, collect the results
in a new list. Written as a loop it takes four lines.

````python
squares = []

for value in range(10):
    squares.append(value ** 2)

print(squares)
````

Python offers a compact form for exactly this, called a **list comprehension**:

```python
[expression for variable in sequence]
```

It reads left to right as "the expression, for each variable in the sequence".

````python
squares = [value ** 2 for value in range(10)]

print(squares)
````

A condition can be added at the end to keep only some of the elements:

```python
[expression for variable in sequence if condition]
```

````python
even_squares = [value ** 2 for value in range(10) if value % 2 == 0]
print(even_squares)

names = ["hydrogen", "helium", "lithium", "beryllium"]
short_names = [name.upper() for name in names if len(name) <= 7]
print(short_names)
````

Comprehensions are used constantly in data analysis code, so it is worth being able to read them.

Keep them short. If a comprehension needs more than one condition or a nested loop, an ordinary
`for` loop is almost always clearer.

# 10. Functions

If you have had some experience with programming you might have noticed that it is common to have to give the same or similar sequences of instructions several times. One possibility would be to simply patiently rewrite everything every time. However apart from being very boring (and a waste of your time), we will see that it is a bad practice (more on this later).

To address these cases (almost) all languages offer the possibility of constructing functions. A programming function is a block of instructions to which a name has been assigned: this allow to simply call the name of the function and the interpreter will understand that it has to execute all the instructions contained within the function.

```python
def function_name():
    statement1
    statement2
    statement3
```

This is indeed a neat trick, but functions are much more flexible. Each function can be defined as accepting parameters from the final user. The instructions within the function will then perform the defined operations according to the passed parameters

```python
def function_name(arg1, arg2,... argN):
    statements_depending_on_arg1_arg2_etc
```

These types of functions are useful to execute some operations. However python functions offer another very interesting possibility: they can return a value. To this respect they can behave as mathematical functions, where the passed parameters are used to calculate the return value.

```python
def function_name(arg1, arg2,... argN):
    statements
    return <value>
```

Functions are very important, not only because they can save your time, but also because they provide a very powerful way to share code. Most programs tend to perform a large number of relatively complex operations. Separating the tasks and wrapping them in functions that can be used upon request is a very important way to manage complexity. This is because we can work a lot on solving a part of the problem we have, wrap it into a function and then simply request the solution without having to worry about or having to remember how it was calculated. This also allows your colleagues to use the function you programmed without having to spend the time to understand how it works.  

To achieve this it is important that every function has a clear behaviour and that that behaviour is well documented. For this python provides syntax to easily store documentation within the function definition:

```python
def function_name(arg1, arg2,... argN):
    ''' Document String'''
    statements
    return <value>
```

If a function provides as document string, it can be seen by calling
```python
help(function_name)
```
Notice that you can see the document string of most functions we will be using in python (for instance you can do `help(print)`).

We will learn what are good practices in producing documentation.

Here is the simplest possible function: it takes no arguments and returns nothing, it just runs the
instructions in its body when called.

Defining a function does **not** run it. The body runs only when the function is called, by writing
its name followed by round brackets.

````python
def greet():
    print("Hello, and welcome to PH6721.")

greet()      # this is the call: without it, nothing is printed
greet()      # a function can be called as many times as we like
````

Functions become useful once they accept **arguments**, so that the same instructions can operate
on different values.

````python
def greet_person(name):
    print(f"Hello {name}, and welcome to PH6721.")

greet_person("Ada")
greet_person("Alan")
````

### Keyword arguments and default values

Both calls above passed their arguments **by position**: the first value goes to the first
parameter, the second to the second. Arguments can also be passed **by name**, and then the order
stops mattering and the call describes itself.

Positional and keyword arguments may be mixed, provided the keyword ones come last.

````python
def area(width, height):
    return width * height

print(area(2, 10))                 # positional: width = 2, height = 10
print(area(height=10, width=2))    # by name: the order no longer matters
print(area(2, height=10))          # mixed, with the keyword argument last
````

A parameter can also be given a **default value**, used whenever the caller leaves it out. This is
how one function covers both the common case and the occasional exception.

````python
def greet(name, greeting="Hello"):        # greeting has a default
    print(f"{greeting} {name}, and welcome to PH6721.")

greet("Ada")                              # the default is used
greet("Alan", "Good morning")             # overridden by position
greet("Grace", greeting="Good evening")   # overridden by name
````

This is why calls to libraries look the way they do throughout the course. In
`plt.plot(x, y, color="red")` and `train_test_split(X, y, test_size=0.2)`, the first arguments are
positional and everything after them is a keyword argument with a sensible default that we have
chosen to override.

## 10.1 The `return` statement

When the function results in some value and that value has to be stored in a variable or needs to be sent back or returned for further operation to the main algorithm, a return statement is used.

````python
def multiply(x, y):
    product = x * y
    return product

result = multiply(4, 5)
print(result)
````

The `multiply` function above accepts two arguments and returns the variable `product`, which
holds their result.

Instead of introducing another variable, the expression itself can be returned. The text in triple
quotes on the first line is a **docstring**: a short description of what the function does.

````python
def multiply(x, y):
    '''Multiply the two input arguments.'''
    return x * y

result = multiply(4, 5)
print(result)
````

Multiple variable can also be returned as a so called tuple. This should be done with care, since this tends to be not very readable when returning many value, and can easily introduce errors when the order of return values is interpreted incorrectly.

````python
def summarize(values):
    highest = max(values)
    lowest = min(values)
    first = values[0]
    last = values[-1]
    return highest, lowest, first, last     # returns a tuple of values

values = [10, 50, 30, 12, 6, 8, 100]

print(summarize(values))

highest, lowest, first, last = summarize(values)     # unpacking the tuple
print(f"highest = {highest}, lowest = {lowest}, first = {first}, last = {last}")
````

## 10.2 Objects, variables, and function arguments

If you are familiar with other programming languages, you might wonder whether Python passes function arguments *by value* or *by reference*.

For Python, this terminology is not very helpful. A better way to think about it is this:

> In Python, variables are names attached to objects.

When we pass a variable to a function, Python passes the object that the variable refers to. Inside the function, the parameter becomes another name for the same object.

This matters because some objects can be modified after they are created. These are called **mutable** objects. Lists and NumPy arrays are mutable.

### Two variables can refer to the same object

Consider the following example:

````python
x = [1, 2, 3]
y = x

y[0] = 100

print(x)
print(y)
````

Changing the list through `y` also changes what we see through `x`, because there is only one list object. Both variables refer to the same object.

If we want an independent list, we must explicitly create a copy.

````python
# Figure code — an illustration, not material.
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch

def name_tag(axis, x, y, text):
    axis.add_patch(FancyBboxPatch((x-0.34,y-0.19), 0.68, 0.38,
                                  boxstyle="round,pad=0.03,rounding_size=0.08",
                                  facecolor="white", edgecolor="0.35", linewidth=1.5))
    axis.text(x, y, text, ha="center", va="center", fontsize=13, fontweight="bold",
              fontfamily="monospace")

def list_block(axis, x, y, entries, colour):
    for k, value in enumerate(entries):
        axis.add_patch(FancyBboxPatch((x+k*0.62-0.29, y-0.29), 0.58, 0.58,
                                      boxstyle="square,pad=0.02",
                                      facecolor=colour, alpha=0.18, edgecolor=colour,
                                      linewidth=1.5))
        axis.text(x+k*0.62, y, str(value), ha="center", va="center", fontsize=12,
                  fontweight="bold")
    axis.text(x+0.62, y-0.55, "one list object", ha="center", va="top", fontsize=9,
              color="0.45", style="italic")

def arrow(axis, start, end, colour="0.45"):
    axis.add_patch(FancyArrowPatch(start, end, arrowstyle="->", mutation_scale=15,
                                   color=colour, linewidth=1.6,
                                   connectionstyle="arc3,rad=0.0"))

fig, axes = plt.subplots(1, 2, figsize=(13, 4.6))

left = axes[0]
name_tag(left, 0.4, 1.55, "x"); name_tag(left, 0.4, 0.45, "y")
list_block(left, 2.0, 1.0, [100, 2, 3], "crimson")
arrow(left, (0.78, 1.5), (1.62, 1.12)); arrow(left, (0.78, 0.5), (1.62, 0.88))
left.set_title("y = x", pad=14, fontfamily="monospace")
left.text(1.7, -0.62, "two names, one object — writing through either\nchanges what the other sees",
          ha="center", va="top", fontsize=10, color="0.3")

right = axes[1]
name_tag(right, 0.4, 1.55, "x"); name_tag(right, 0.4, 0.45, "y")
list_block(right, 2.0, 1.65, [1, 2, 3], "tab:blue")
list_block(right, 2.0, 0.35, [100, 2, 3], "tab:blue")
arrow(right, (0.78, 1.58), (1.62, 1.65)); arrow(right, (0.78, 0.42), (1.62, 0.35))
right.set_title("y = x.copy()", pad=14, fontfamily="monospace")
right.text(1.7, -0.62, "two names, two objects — writing through one\nleaves the other alone",
           ha="center", va="top", fontsize=10, color="0.3")

for axis in axes:
    axis.set_xlim(-0.2, 3.6); axis.set_ylim(-0.75, 2.3); axis.axis("off")
fig.suptitle("A variable is a name attached to an object", fontsize=14, fontweight="bold")
plt.show()
````

````python
x = [1, 2, 3]
y = x.copy()

y[0] = 100

print(x)
print(y)
````

Now `x` and `y` refer to two different list objects.

### Modifying an object inside a function

The same idea applies to function arguments. If the object is mutable, the function can modify it.

````python
def modify_list(L):
    L[0] = 999

numbers = [1, 2, 3]

modify_list(numbers)

print(numbers)
````

The function did not receive a separate copy of the list. The parameter `L` referred to the same list object as `numbers`.

### Reassigning a variable is different

Now compare the previous example with this one:

````python
def replace_list(L):
    L = [999, 999, 999]

numbers = [1, 2, 3]

replace_list(numbers)

print(numbers)
````

Here the original list is unchanged. Inside the function, `L` is reassigned to a new list. This does not change what `numbers` refers to outside the function.

So the important distinction is:

- modifying an existing object can affect the caller;
- reassigning a local variable inside the function does not affect the caller.


### A dangerous example

Let us now look at a function that computes the sum of the squares of a list of numbers. The following version works numerically, but it has a serious problem.

````python
def sum_squares(values):
    for i in range(len(values)):
        values[i] *= values[i]    # This modifies the input object!
        
    summation = 0
    for elem in values:
        summation += elem
        
    return summation
    
print(sum_squares([2, 3, 5, 12, 4]))
````

The problem becomes visible when the input list already exists before calling the function.

````python
my_numbers = [2, 3, 5, 12, 4]

print(sum_squares(my_numbers))
print(my_numbers)
````

The function has changed `my_numbers`. This is usually not what we want from a function whose purpose is only to compute a value.

### Safer version: copy before modifying

If a function needs to modify the data internally, but the original object should remain unchanged, make an explicit copy.

````python
def sum_squares(values):
    local_copy = values.copy()     # Make an independent copy
    
    for i in range(len(local_copy)):
        local_copy[i] *= local_copy[i]
        
    summation = 0
    for elem in local_copy:
        summation += elem
        
    return summation

my_numbers = [2, 3, 5, 12, 4]

print(sum_squares(my_numbers))
print(my_numbers)
````

Now the function returns the sum of squares, but the original list is not modified.

### NumPy arrays behave similarly

NumPy arrays are also mutable. This is very important in scientific computing and deep learning, because arrays and tensors are often large objects.

````python
import numpy as np

x = np.array([1, 2, 3])
y = x

y[0] = 100

print(x)
print(y)
````

Again, `x` and `y` refer to the same object. To create an independent array, use `.copy()`.

````python
x = np.array([1, 2, 3])
y = x.copy()

y[0] = 100

print(x)
print(y)
````

### `.copy()` copies one level only

A copy of a list is a new list, but it holds the *same* objects the original held. If those objects
are themselves mutable — a list of lists, a dictionary of arrays — then writing into one of them is
visible through both copies.

````python
import copy

nested = [[1, 2], [3, 4]]

shallow = nested.copy()          # a new outer list, holding the same two inner lists
shallow[0][0] = 99
print("after writing into the shallow copy:", nested)

nested = [[1, 2], [3, 4]]
deep = copy.deepcopy(nested)     # copies the inner lists too
deep[0][0] = 99
print("after writing into the deep copy:   ", nested)
````

### Rule of thumb

- If your function is supposed to compute and return a value, avoid modifying its input arguments.
- If you need to modify an input internally, first make an explicit copy.
- If your function is supposed to modify an object, make that clear in the function name and documentation.
- Remember: assignment such as `y = x` does not copy the object. It only creates another name for the same object.
- `.copy()` copies one level only, as the cells just above show. If the object contains other mutable objects — a list of lists, say — those inner objects are still shared. `copy.deepcopy` is the tool for that case.

This distinction is especially important later, when working with NumPy arrays, Pandas DataFrames, and deep-learning tensors.


## 10.3 Global and local variables

A variable created inside a function is usually **local**: it exists only while that function is running.

A variable created outside all functions is **global**: it can be read from many parts of the notebook.

For small examples, global variables can be convenient. In real programs and machine-learning notebooks, however, global state can easily create confusing bugs: a function may depend on a value that was changed in a different cell, or a notebook may stop working when cells are run in a different order.

A good rule is:

- use global variables mainly for constants, such as configuration values;
- pass information into functions through arguments;
- return results explicitly;
- avoid functions that silently modify global state.

In the example below, the inner function defines its own `x`. That `x` is local to the inner
function, so the outer function's `x` is untouched.

````python
def outer_function():
    x = 1

    def inner_function():
        x = 2
        print("Inside inner_function, x =", x)

    inner_function()
    print("Inside outer_function, x =", x)

outer_function()
````

The next example shows why global variables need care. The keyword `global` tells Python that the
function should modify the *global* variable `x`, rather than create a local one.

This is occasionally useful, but it makes the function less self-contained and harder to reuse:
calling it changes state somewhere else in the program.

````python
x = 0.0     # a global variable

def outer_function():
    x = 1.0                 # local to outer_function

    def inner_function():
        global x            # refers to the global x, not to outer_function's
        x = 2.0
        print("Inside inner_function, x =", x)

    inner_function()
    print("Inside outer_function, x =", x)

outer_function()
print("The global x is now", x)
````

## 10.4 Lambda functions

A **lambda function** is a short anonymous function: a function defined directly where it is needed, rather than with a separate `def` block.

The syntax is:

```python
lambda arguments: expression
```

Lambda functions are useful when a function is needed only once, especially when another function expects a function as one of its arguments. This happens often in data analysis and deep-learning code, for instance when sorting, transforming datasets, or defining small preprocessing operations.

````python
square = lambda x: x**2

square(8)
````

The lambda function above is equivalent to the following ordinary function definition:

````python
def square(x):
    return x**2

square(8)
````

The main advantage of lambda functions is that they can be written directly inside another function call. For example, the function `sorted(...)` can receive a `key` function telling Python what criterion should be used for sorting.

````python
pairs = [(1, 3), (2, 1), (4, 2)]

# Sort the pairs according to their second element.
sorted(pairs, key=lambda x: x[1])
````

A good rule of thumb is:

- use `def` for functions that are important, reused, or need a clear name;
- use `lambda` for very short functions that are passed as arguments and used only locally.

Avoid writing complicated lambda functions. If the expression is difficult to read, define a normal function instead.

## 11.1 Basic plotting<a name="basic_plotting"></a>

Here is a simple example, which plots the function $y = \sin(x)$.

````python
import numpy as np
import matplotlib.pyplot as plt

# Plot sin(x) over a reasonable x range.
x = np.linspace(-6., 6., 1000)
y = np.sin(x)

plt.plot(x, y)

# Format the figure nicely.
plt.xlabel('x')
plt.ylabel('y')
plt.title('y = sin(x)')
plt.ylim(-1.1, 1.1);
````

Two details in that cell.

The trailing semicolon in `plt.ylim(-1.1, 1.1);` has nothing to do with separating statements as in
section 2. In a notebook it suppresses **whatever the last line evaluates to**, so that it is not
displayed above the figure. Here that value is the pair of limits, `(-1.1, 1.1)`; had the last line
been `plt.plot(...)`, it would have been the `<matplotlib...>` line you may have seen elsewhere. The usual alternative is to
finish with `plt.show()`, which is what `00A` and the Class 1 notebooks do.

`plt.plot` also accepts a single list, in which case the values are plotted against their position

# 13. Good programming practice

- **Give meaningful names to everything.** Really, I mean it. `n` and `tmp` will be meaningless to
  you in a fortnight.
- **Follow the standard Python naming conventions.** Python's style guide, PEP 8, specifies:

| Kind | Convention | Example |
|---|---|---|
| Variables and functions | lower case with underscores | `learning_rate`, `train_model(...)` |
| Classes and types | capitalised words, no underscores | `NeuralNetwork` |
| Constants | upper case with underscores | `MAX_ITERATIONS` |

  This is the convention used by every library in this course — `np.mean`, `df.head`,
  `model.fit`, `torch.no_grad` — so following it makes your code look like the code around it.
  Other languages use different conventions; if you have written C++ or Java you may be used to
  `camelCase`, but in Python it will look out of place.

- **Comment your code**, and say *why*, not *what*. The code already says what it does.
- **Do not repeat yourself.** If you have copied and pasted a block twice, it should be a function.
- **Do not reinvent the wheel** (unless I say so ;) ). If NumPy already has it, use NumPy's.

# Key Takeaways

- A notebook runs top to bottom; a cell can only use what earlier cells have created.
- Python infers types from the values assigned. It will also let you change a variable's type,
  which is legal and almost always a bad idea.
- f-strings are the modern way to build strings: `f"{value:.3f}"`.
- Four containers cover most needs: `list` for ordered sequences, `tuple` for fixed groups,
  `dict` for lookup by name, `set` for distinct values.
- Indentation defines blocks in Python. It is syntax, not decoration.
- List comprehensions express "transform each element" compactly. Keep them short.
- Defining a function does not run it; calling it does.
- `y = x` does not copy an object, it gives it a second name. Use `.copy()` when you need an
  independent object — this matters enormously for NumPy arrays and tensors later.
- Follow PEP 8 naming: `snake_case` for variables and functions, `CapWords` for classes.
